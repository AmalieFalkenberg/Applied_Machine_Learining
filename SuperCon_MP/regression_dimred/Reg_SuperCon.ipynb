```python
from __future__ import print_function, division
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import lightgbm as lgb
import shap
import optuna

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, root_mean_squared_error
from sklearn.inspection import permutation_importance
from lightgbm import early_stopping

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "font.size": 16,
    "axes.grid": True,
    "grid.alpha": 0.2
})

# ── 1. Load data ───────────────────────────────────────────────────────────────
print("=== Loading Data ===\n")

df_wchem = pd.read_csv('SuperCon_wChem.csv')
tc = np.load('/Users/fbnielsen/Desktop/AML/Final project/Supplementary_1/SuperCon_critical_temperatures.npy')


print(f"Samples: {len(tc)}")
print(f"Features:        {df_wchem.shape[1]}")

# ── 2. Data quality ────────────────────────────────────────────────────────────
print("\n=== Data Quality Checks ===\n")

# Convert booleans to int
bool_cols = df_wchem.select_dtypes(include='bool').columns
df_wchem[bool_cols] = df_wchem[bool_cols].astype(int)

# Keep only numeric columns
df_wchem = df_wchem.select_dtypes(include='number')

# Fill NaNs with median
nan_total = df_wchem.isna().sum().sum()
if nan_total:
    print(f"Filling {nan_total} NaN values with column median")
    df_wchem = df_wchem.fillna(df_wchem.median())

# Remove zero-variance columns
zero_var = [c for c in df_wchem.columns if df_wchem[c].nunique() <= 1]
if zero_var:
    print(f"Dropping {len(zero_var)} zero-variance columns")
    df_wchem = df_wchem.drop(columns=zero_var)

print(f"✅ Clean feature matrix: {df_wchem.shape}")

# ── 3. Remove highly correlated features (>0.90) ─────────────────────────────
print("\n=== Removing Highly Correlated Features (threshold=0.90) ===\n")

corr_matrix   = df_wchem.corr(method='pearson').abs()
upper         = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop_corr  = [c for c in upper.columns if any(upper[c] > 0.90)]
df_wchem      = df_wchem.drop(columns=to_drop_corr)

print(f"Dropped {len(to_drop_corr)} correlated features. Remaining: {df_wchem.shape[1]}")

# ── 4. Train / Val / Test split (60 / 20 / 20) ────────────────────────────────
print("\n=== Train / Val / Test Split (60 / 20 / 20) ===\n")

Xr_temp, Xr_test, Yr_temp, Yr_test = train_test_split(
    df_wchem, tc, test_size=0.20, random_state=42
)
Xr_train, Xr_val, Yr_train, Yr_val = train_test_split(
    Xr_temp, Yr_temp, test_size=0.25, random_state=42   # 0.25 × 0.80 = 0.20 total
)

Yr_train = Yr_train.squeeze()
Yr_val   = Yr_val.squeeze()
Yr_test  = Yr_test.squeeze()

print(f"Train: {len(Xr_train)}  |  Val: {len(Xr_val)}  |  Test: {len(Xr_test)}")

# ── 5. Baseline model (all features, log-Tc) ──────────────────────────────────
print("\n=== Baseline Model ===\n")

Yr_train_log = np.log1p(Yr_train)
Yr_test_log  = np.log1p(Yr_test)
Yr_val_log   = np.log1p(Yr_val)

baseline_params = {
    'boosting_type': 'gbdt',
    'objective':     'regression',
    'num_leaves':    31,
    'learning_rate': 0.05,
    'n_estimators':  1000,
    'verbosity':     -1,
    'random_state':  42,
    'n_jobs':        -1,
}

gbm_base = lgb.LGBMRegressor(**baseline_params)
gbm_base.fit(
    Xr_train, Yr_train_log,
    eval_set=[(Xr_test, Yr_test_log)],
    eval_metric='rmse',
    callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
)

y_pred_base = np.expm1(gbm_base.predict(Xr_val))
rmse_base   = root_mean_squared_error(Yr_val, y_pred_base)
r2_base     = r2_score(Yr_val, y_pred_base)
print(f"Baseline — RMSE: {rmse_base:.3f} K   R²: {r2_base:.3f}  ({Xr_train.shape[1]} features)")

# ─────────────────────────────────────────────────────────────────────────────
# ── 6. FEATURE SELECTION — THREE METHODS ─────────────────────────────────────
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "="*60)
print("  FEATURE SELECTION")
print("="*60)

# ── 6a. Method 1: SHAP importance ─────────────────────────────────────────────
print("\n--- Method 1: SHAP Importance ---\n")

Xr_test_shap = Xr_test.astype(float)
explainer    = shap.TreeExplainer(gbm_base)
shap_values  = explainer.shap_values(Xr_test_shap)

shap_importance = pd.Series(
    np.abs(shap_values).mean(axis=0),
    index=Xr_train.columns
).sort_values(ascending=False)

print("Top 15 SHAP features:")
for feat, val in shap_importance.head(15).items():
    print(f"  {feat:<50} {val:.5f}")

# SHAP summary plot
shap.summary_plot(shap_values, Xr_test_shap,
                  feature_names=Xr_test_shap.columns,
                  plot_type="violin", max_display=20, show=False)
plt.title("SHAP Feature Importance (Baseline Model)", fontsize=13)
plt.tight_layout()
plt.savefig('shap_summary_baseline.png', dpi=150, bbox_inches='tight')
plt.show()

# ── 6b. Method 2: Permutation importance ──────────────────────────────────────
print("\n--- Method 2: Permutation Importance ---\n")

perm_result = permutation_importance(
    gbm_base, Xr_val, Yr_val_log,
    n_repeats=10,
    random_state=42,
    n_jobs=-1,
    scoring='r2',
)

perm_importance = pd.Series(
    perm_result.importances_mean,
    index=Xr_train.columns
).sort_values(ascending=False)

print("Top 15 Permutation importance features:")
for feat, val in perm_importance.head(15).items():
    print(f"  {feat:<50} {val:.5f}")

# ── 6c. Method 3: LightGBM native gain importance ────────────────────────────
print("\n--- Method 3: LightGBM Native Gain Importance ---\n")

lgb_importance = pd.Series(
    gbm_base.booster_.feature_importance(importance_type='gain'),
    index=Xr_train.columns
).sort_values(ascending=False)

print("Top 15 LGB gain importance features:")
for feat, val in lgb_importance.head(15).items():
    print(f"  {feat:<50} {val:.1f}")

# ── 6d. Combine: union of top-N from each method ──────────────────────────────
print("\n--- Combined Feature Set ---\n")

N = 20   # top-N from each method
top_shap  = set(shap_importance.head(N).index)
top_perm  = set(perm_importance.head(N).index)
top_lgb   = set(lgb_importance.head(N).index)

# Consensus: features appearing in at least 2 out of 3 methods
all_candidates = top_shap | top_perm | top_lgb
consensus_feats = sorted([
    f for f in all_candidates
    if sum([f in top_shap, f in top_perm, f in top_lgb]) >= 2
])

print(f"Top-{N} SHAP:        {len(top_shap)} features")
print(f"Top-{N} Permutation: {len(top_perm)} features")
print(f"Top-{N} LGB gain:    {len(top_lgb)} features")
print(f"Consensus (≥2/3):    {len(consensus_feats)} features")
print("\nConsensus feature list:")
for f in consensus_feats:
    in_methods = []
    if f in top_shap: in_methods.append("SHAP")
    if f in top_perm: in_methods.append("Perm")
    if f in top_lgb:  in_methods.append("LGB")
    print(f"  {f:<50} [{', '.join(in_methods)}]")

# ── 6e. Feature count sweep — find elbow point ────────────────────────────────
print("\n--- Feature Count Sweep (elbow point) ---\n")

sweep_results = {}
n_sweep = [5, 10, 15, 20, 25, 30, 40, 50, len(consensus_feats), Xr_train.shape[1]]
n_sweep = sorted(set(n for n in n_sweep if n <= Xr_train.shape[1]))

for n_feats in n_sweep:
    # Use SHAP ranking for the sweep (most stable)
    feats = shap_importance.head(n_feats).index.tolist()
    m = lgb.LGBMRegressor(**baseline_params)
    m.fit(
        Xr_train[feats], Yr_train_log,
        eval_set=[(Xr_test[feats], Yr_test_log)],
        eval_metric='rmse',
        callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
    )
    y_pred_sweep = np.expm1(m.predict(Xr_val[feats]))
    sweep_results[n_feats] = {
        'rmse': root_mean_squared_error(Yr_val, y_pred_sweep),
        'r2':   r2_score(Yr_val, y_pred_sweep),
    }

print(f"  {'N':>5}  RMSE (K)    R²")
print(f"  {'─'*30}")
for n, metrics in sweep_results.items():
    marker = ' ◄ baseline' if n == Xr_train.shape[1] else ''
    print(f"  {n:>5}  {metrics['rmse']:>8.3f}   {metrics['r2']:.3f}{marker}")

# Plot sweep
fig, ax1 = plt.subplots(figsize=(9, 5))
ax2 = ax1.twinx()
ns   = list(sweep_results.keys())
rmse_vals = [sweep_results[n]['rmse'] for n in ns]
r2_vals   = [sweep_results[n]['r2']   for n in ns]
ax1.plot(ns, rmse_vals, 'b-o', label='RMSE')
ax2.plot(ns, r2_vals,   'r-s', label='R²')
ax1.set_xlabel('Number of features')
ax1.set_ylabel('RMSE (K)', color='b')
ax2.set_ylabel('R²', color='r')
ax1.tick_params(axis='y', colors='b')
ax2.tick_params(axis='y', colors='r')
ax1.set_title('Feature count vs performance (SHAP ranking)')
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='center right')
plt.tight_layout()
plt.savefig('feature_sweep.png', dpi=150, bbox_inches='tight')
plt.show()

# Pick the feature set for final model — use consensus or best from sweep
best_n    = min(sweep_results, key=lambda n: sweep_results[n]['rmse'])
feat_shap = shap_importance.head(best_n).index.tolist()
# Override with consensus if it performs similarly (within 5% RMSE)
if sweep_results.get(len(consensus_feats), {}).get('rmse', 9e9) \
        <= sweep_results[best_n]['rmse'] * 1.05:
    feat_shap = consensus_feats
    print(f"\n✅ Using consensus feature set ({len(feat_shap)} features)")
else:
    print(f"\n✅ Using top-{best_n} SHAP features (best sweep RMSE)")

# ─────────────────────────────────────────────────────────────────────────────
# ── 7. OPTUNA HYPERPARAMETER OPTIMISATION ────────────────────────────────────
# ─────────────────────────────────────────────────────────────────────────────
print("\n=== Optuna Hyperparameter Optimisation ===\n")

def objective_shap(trial):
    lr      = trial.suggest_float("learning_rate", 1e-4, 0.05, log=True)
    patience = int(min(max(50, 0.1 / lr), 500))

    params = {
        "objective":         "regression",
        "boosting_type":     "gbdt",
        "verbosity":         -1,
        "random_state":      42,
        "n_jobs":            -1,
        "learning_rate":     lr,
        "num_leaves":        trial.suggest_int  ("num_leaves",        16,  256),
        "max_depth":         trial.suggest_int  ("max_depth",          3,   12),
        "min_child_samples": trial.suggest_int  ("min_child_samples",  5,  200),
        "subsample":         trial.suggest_float("subsample",         0.5,  1.0),
        "colsample_bytree":  trial.suggest_float("colsample_bytree",  0.5,  1.0),
        "reg_alpha":         trial.suggest_float("reg_alpha",        1e-8, 10.0, log=True),
        "reg_lambda":        trial.suggest_float("reg_lambda",       1e-8, 10.0, log=True),
    }

    model = lgb.LGBMRegressor(**params, n_estimators=10000)
    model.fit(
        Xr_train[feat_shap], Yr_train_log,
        eval_set=[(Xr_test[feat_shap], Yr_test_log)],
        eval_metric="rmse",
        callbacks=[lgb.early_stopping(patience), lgb.log_evaluation(0)],
    )
    y_pred = np.expm1(model.predict(Xr_test[feat_shap]))
    return root_mean_squared_error(Yr_test, y_pred)

optuna.logging.set_verbosity(optuna.logging.WARNING)
study_shap = optuna.create_study(
    direction="minimize",
    sampler=optuna.samplers.TPESampler(seed=42),
    pruner=optuna.pruners.MedianPruner(n_startup_trials=10, n_warmup_steps=2),
)
study_shap.optimize(objective_shap, n_trials=100, show_progress_bar=True)

print(f"\nBest RMSE (test): {study_shap.best_value:.3f} K")
print("Best parameters:")
for k, v in study_shap.best_params.items():
    print(f"  {k}: {v}")

# ── 8. Retrain best model and evaluate ────────────────────────────────────────
print("\n=== Final Model Evaluation ===\n")

best_p   = study_shap.best_params
patience = int(min(max(50, 0.1 / best_p['learning_rate']), 500))

lgb_train_ds = lgb.Dataset(Xr_train[feat_shap], np.log1p(Yr_train))
lgb_test_ds  = lgb.Dataset(Xr_test[feat_shap],  np.log1p(Yr_test), reference=lgb_train_ds)

full_params = {**best_p, 'objective': 'regression', 'boosting_type': 'gbdt',
               'verbosity': -1, 'random_state': 42}

gbm_final = lgb.train(
    full_params,
    lgb_train_ds,
    num_boost_round=5000,
    valid_sets=lgb_test_ds,
    callbacks=[early_stopping(patience), lgb.log_evaluation(0)],
)

y_pred_val = np.expm1(gbm_final.predict(Xr_val[feat_shap]))
rmse_final = root_mean_squared_error(Yr_val, y_pred_val)
mae_final  = mean_absolute_error(Yr_val, y_pred_val)
r2_final   = r2_score(Yr_val, y_pred_val)

print(f"Validation performance ({len(feat_shap)} features):")
print(f"  RMSE : {rmse_final:.4f} K")
print(f"  MAE  : {mae_final:.4f} K")
print(f"  R²   : {r2_final:.4f}")
print(f"\nImprovement over baseline:")
print(f"  RMSE : {rmse_base:.3f} → {rmse_final:.3f} K  ({(rmse_base-rmse_final)/rmse_base*100:+.1f}%)")
print(f"  R²   : {r2_base:.3f} → {r2_final:.3f}")

# ── 9. Plots ───────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(13, 5), dpi=100)

for ax, y_pred, title in zip(
    axes,
    [y_pred_base, y_pred_val],
    [f'Baseline ({Xr_train.shape[1]} feats)  R²={r2_base:.3f}',
     f'Optimised ({len(feat_shap)} feats)  R²={r2_final:.3f}']
):
    ax.scatter(Yr_val, y_pred, s=2, alpha=0.5)
    lims = [min(Yr_val.min(), np.array(y_pred).min()),
            max(Yr_val.max(), np.array(y_pred).max())]
    ax.plot(lims, lims, 'r--', lw=1)
    ax.set_xlabel('True Tc (K)')
    ax.set_ylabel('Predicted Tc (K)')
    ax.set_title(title, fontsize=13)
    ax.set_xscale('log')
    ax.set_yscale('log')

plt.suptitle('Cuprate Tc Regression — Baseline vs Optimised', fontsize=14)
plt.tight_layout()
plt.savefig('regression_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

# ── 10. SHAP on final model ────────────────────────────────────────────────────
print("\n=== SHAP Summary (Final Model) ===\n")

explainer_final = shap.TreeExplainer(gbm_final)
shap_final      = explainer_final.shap_values(Xr_val[feat_shap].astype(float))

shap.summary_plot(shap_final, Xr_val[feat_shap].astype(float),
                  feature_names=feat_shap,
                  plot_type="violin", max_display=20, show=False)
plt.title("SHAP Feature Importance (Optimised Model)", fontsize=13)
plt.tight_layout()
plt.savefig('shap_summary_final.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nSaved: shap_summary_baseline.png, feature_sweep.png,")
print("       regression_comparison.png, shap_summary_final.png")
```

```python
import pandas as pd
import numpy as np
import umap
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.colors as mcolors
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from scipy.stats import spearmanr

# ── 1. Load data ───────────────────────────────────────────────────────────────
df = df_wchem #pd.read_csv('/Users/fbnielsen/Desktop/AML/Final project/Supplementary_1/SuperCon_features.csv')
print(f"Loaded: {df.shape[0]} rows × {df.shape[1]} columns")

feature_cols = df.columns.tolist()   # all 147 columns are numeric features
X = df[feature_cols].copy()

# ── 2. Scale ───────────────────────────────────────────────────────────────────
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_wchem)

# ── 3. Run UMAP ────────────────────────────────────────────────────────────────
print("Running UMAP (this may take a minute on 13k points)...")
reducer = umap.UMAP(n_components=2, random_state=42,n_neighbors=180, min_dist=0.26)
embedding = reducer.fit_transform(X_scaled)
umap1, umap2 = embedding[:, 0], embedding[:, 1]
print("UMAP done.")

# ── 4. Custom colormap: jet from 1/3 ──────────────────────────────────────────
jet_trimmed = mcolors.LinearSegmentedColormap.from_list(
    'jet_trimmed', plt.cm.jet(np.linspace(1/3, 1.0, 256))
)

# ── 5. Spearman correlations ───────────────────────────────────────────────────
print("Computing Spearman correlations...")
corr_results = []
for col in feature_cols:
    r1, _ = spearmanr(X[col].values, umap1)
    r2, _ = spearmanr(X[col].values, umap2)
    corr_results.append({'feature': col, 'corr_umap1': r1, 'corr_umap2': r2})

corr_df = pd.DataFrame(corr_results).set_index('feature')

print("\nTop 10 correlates with UMAP axis 1:")
for feat, val in corr_df['corr_umap1'].abs().sort_values(ascending=False).head(10).items():
    print(f"  {feat:<55} {val:.4f}")

print("\nTop 10 correlates with UMAP axis 2:")
for feat, val in corr_df['corr_umap2'].abs().sort_values(ascending=False).head(10).items():
    print(f"  {feat:<55} {val:.4f}")

# ── 6. Random Forest feature importance ───────────────────────────────────────
print("\nFitting Random Forests (200 trees each)...")
rf1 = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
rf1.fit(X_scaled, umap1)
imp1 = pd.Series(rf1.feature_importances_, index=feature_cols).sort_values(ascending=False)

rf2 = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
rf2.fit(X_scaled, umap2)
imp2 = pd.Series(rf2.feature_importances_, index=feature_cols).sort_values(ascending=False)

print(f"\nR² for UMAP1 prediction: {rf1.score(X_scaled, umap1):.3f}")
print(f"R² for UMAP2 prediction: {rf2.score(X_scaled, umap2):.3f}")

print("\nTop 10 features driving UMAP axis 1:")
for feat, val in imp1.head(10).items():
    bar = '█' * int(val * 300)
    print(f"  {feat:<55} {val:.4f}  {bar}")

print("\nTop 10 features driving UMAP axis 2:")
for feat, val in imp2.head(10).items():
    bar = '█' * int(val * 300)
    print(f"  {feat:<55} {val:.4f}  {bar}")

# ── 7. Plots ───────────────────────────────────────────────────────────────────
def scatter_feature(ax, feat_vals, label, cmap=jet_trimmed):
    sc = ax.scatter(umap1, umap2, c=feat_vals, cmap=cmap, s=3, alpha=0.5,
                    vmin=np.percentile(feat_vals, 1),
                    vmax=np.percentile(feat_vals, 99))
    plt.colorbar(sc, ax=ax, pad=0.02)
    ax.set_title(label, fontsize=9)
    ax.set_xlabel('UMAP 1', fontsize=8)
    ax.set_ylabel('UMAP 2', fontsize=8)
    ax.tick_params(labelsize=7)

fig = plt.figure(figsize=(22, 18))
gs = gridspec.GridSpec(5, 5, figure=fig, hspace=0.45, wspace=0.38)

# Row 0: top 3 RF drivers of axis 1
for i, feat in enumerate(imp1.head(5).index):
    ax = fig.add_subplot(gs[0, i])
    scatter_feature(ax, X[feat].values, f"UMAP1 driver #{i+1}:\n{feat}")

# Row 1: top 3 RF drivers of axis 2
for i, feat in enumerate(imp2.head(5).index):
    ax = fig.add_subplot(gs[1, i])
    scatter_feature(ax, X[feat].values, f"UMAP2 driver #{i+1}:\n{feat}")

# Row 2: feature importance bar charts
ax1 = fig.add_subplot(gs[2, :2])
imp1.head(15).plot.barh(ax=ax1)
ax1.invert_yaxis()
ax1.set_title('Feature importance → UMAP axis 1', fontsize=10)
ax1.set_xlabel('Importance', fontsize=8)
ax1.tick_params(labelsize=7)

ax2 = fig.add_subplot(gs[2, 3:])
imp2.head(15).plot.barh(ax=ax2)
ax2.invert_yaxis()
ax2.set_title('Feature importance → UMAP axis 2', fontsize=10)
ax2.set_xlabel('Importance', fontsize=8)
ax2.tick_params(labelsize=7)

plt.suptitle('UMAP of SuperCon MAGPIE features — axis interpretation', fontsize=13, y=1.01)
plt.savefig('umap_supercon_interpretation.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved: umap_supercon_interpretation.png")



# # ── 9. Save results ────────────────────────────────────────────────────────────
# summary = pd.DataFrame({
#     'RF_importance_axis1': imp1,
#     'RF_importance_axis2': imp2,
#     'Spearman_axis1':      corr_df['corr_umap1'],
#     'Spearman_axis2':      corr_df['corr_umap2'],
# }).round(4).sort_values('RF_importance_axis1', ascending=False)
# summary.to_csv('umap_supercon_feature_importance.csv')
# print("Saved: umap_supercon_feature_importance.csv")

# out = pd.DataFrame({'umap_x': umap1, 'umap_y': umap2})
# out.to_csv('umap_supercon_embedding.csv', index=False)
# print("Saved: umap_supercon_embedding.csv")
```
