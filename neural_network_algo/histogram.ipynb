```python
import matplotlib.pyplot as plt
import numpy as np

# Data extracted from the image
t_limits = ['5', '30', '77']
roc_auc_scores = [0.9727, 0.9877, 0.9889]
accuracies = [92.32, 95.39, 96.12]

# Set up the figure with two subplots side-by-side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
fig.suptitle('Classification Results by $T_{c.limit}$', fontsize=16)

# Plot 1: ROC-AUC Bar Chart
bars1 = ax1.bar(t_limits, roc_auc_scores, color='skyblue', edgecolor='black')
ax1.set_title('ROC-AUC Score')
ax1.set_xlabel('$T_{c.limit}$')
ax1.set_ylabel('ROC-AUC')
ax1.set_ylim(0.95, 1.0) # Adjusted for better visualization of differences

# Add value labels on top of the bars
for bar in bars1:
    yval = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2, yval + 0.001, round(yval, 4), ha='center', va='bottom')

# Plot 2: Accuracy Bar Chart
bars2 = ax2.bar(t_limits, accuracies, color='lightgreen', edgecolor='black')
ax2.set_title('Accuracy (%)')
ax2.set_xlabel('$T_{c.limit}$')
ax2.set_ylabel('Accuracy')
ax2.set_ylim(90, 100) # Adjusted for better visualization of differences

# Add value labels on top of the bars
for bar in bars2:
    yval = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2, yval + 0.1, f"{round(yval, 2)}%", ha='center', va='bottom')

# Adjust layout to prevent overlap and display the plot
plt.tight_layout()
plt.show()
```
