```python
# ══════════════════════════════════════════════════════════════════════════════
# ── 0. Imports ─────────────────────────────────────────────────────────────────
# ══════════════════════════════════════════════════════════════════════════════
import pandas as pd
import numpy as np
import shap
import xgboost as xgb
import lightgbm as lgb
import optuna
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import (log_loss, roc_auc_score, accuracy_score,
                             classification_report, confusion_matrix)

plt.rcParams.update({
    "font.family": "serif", "font.serif": ["Times New Roman"],
    "font.size": 16, "axes.grid": True, "grid.alpha": 0.2
})

# ══════════════════════════════════════════════════════════════════════════════
# ── 1. Load data ───────────────────────────────────────────────────────────────
# ══════════════════════════════════════════════════════════════════════════════
df_wchem = pd.read_csv('/Users/fbnielsen/Desktop/3DSC/df_wchem.csv').drop('Unnamed: 0', axis=1)
tc       = pd.read_csv('/Users/fbnielsen/Desktop/3DSC/tc_3dsc.csv').drop('Unnamed: 0', axis=1)

tc_series = tc.squeeze().reset_index(drop=True)
df_wchem  = df_wchem.reset_index(drop=True)

# Binary label: 1 = superconductor (tc > 0), 0 = non-superconductor
label = (tc_series > 0).astype(int)
label.name = 'is_sc'

print(f"Dataset:             {df_wchem.shape}")
print(f"Superconductors:     {label.sum()} ({label.mean()*100:.1f}%)")
print(f"Non-superconductors: {(1-label).sum()} ({(1-label.mean())*100:.1f}%)")

# ══════════════════════════════════════════════════════════════════════════════
# ── 2. Data quality ────────────────────────────────────────────────────────────
# ══════════════════════════════════════════════════════════════════════════════
feature_cols = df_wchem.columns.tolist()

zero_var = [c for c in feature_cols if df_wchem[c].nunique() <= 1]
if zero_var:
    df_wchem.drop(columns=zero_var, inplace=True)
    feature_cols = df_wchem.columns.tolist()
    print(f"Dropped {len(zero_var)} zero-variance columns")

df_wchem[feature_cols] = df_wchem[feature_cols].fillna(df_wchem[feature_cols].median())

corr    = df_wchem[feature_cols].corr().abs()
upper   = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
to_drop = [c for c in upper.columns if any(upper[c] > 0.90)]
df_wchem.drop(columns=to_drop, inplace=True)
feature_cols = df_wchem.columns.tolist()
print(f"Features after correlation filter: {len(feature_cols)}")

# ══════════════════════════════════════════════════════════════════════════════
# ── 3. Train / Val / Test split (60 / 20 / 20) — mirrors regression notebook ──
# ══════════════════════════════════════════════════════════════════════════════
df_all = df_wchem.copy()
df_all['is_sc'] = label

X_train_all, x_test, Y_train_all, y_test = train_test_split(
    df_wchem, label, test_size=0.4, random_state=42, stratify=label)
X_test, X_val, y_test, y_val = train_test_split(
    x_test, y_test, test_size=0.5, random_state=42, stratify=y_test)

print(f"Train: {len(X_train_all)}  |  Test: {len(X_test)}  |  Val: {len(X_val)}")

# ══════════════════════════════════════════════════════════════════════════════
# ── 4. Initial LightGBM — all features ────────────────────────────────────────
# ══════════════════════════════════════════════════════════════════════════════
imbalance_ratio = (Y_train_all == 0).sum() / (Y_train_all == 1).sum()
print(f"\nClass imbalance ratio (scale_pos_weight): {imbalance_ratio:.2f}")

init_params = {
    'boosting_type': 'gbdt',
    'objective':     'binary',
    'num_leaves':    31,
    'verbose':       -1,
    'scale_pos_weight': imbalance_ratio,
}

lgb_train_ds = lgb.Dataset(X_train_all, Y_train_all)
lgb_eval_ds  = lgb.Dataset(X_test,      y_test, reference=lgb_train_ds)

gbm_init = lgb.train(
    init_params,
    lgb_train_ds,
    num_boost_round=1000,
    valid_sets=lgb_eval_ds,
    callbacks=[lgb.early_stopping(20), lgb.log_evaluation(50)],
)

y_pred_val_prob = gbm_init.predict(X_val, num_iteration=gbm_init.best_iteration)
y_pred_val      = (y_pred_val_prob > 0.5).astype(int)

print("\nInitial model — validation set")
print(f"  ROC-AUC:  {roc_auc_score(y_val, y_pred_val_prob):.4f}")
print(f"  Accuracy: {accuracy_score(y_val, y_pred_val)*100:.2f}%")
print(f"  LogLoss:  {log_loss(y_val, y_pred_val_prob):.4f}")

# ══════════════════════════════════════════════════════════════════════════════
# ── 5. SHAP feature importance ─────────────────────────────────────────────────
# ══════════════════════════════════════════════════════════════════════════════
X_test_shap = X_test.copy()
bool_cols = X_test_shap.select_dtypes(include='bool').columns
X_test_shap[bool_cols] = X_test_shap[bool_cols].astype(int)

explainer   = shap.TreeExplainer(gbm_init)
shap_values = explainer.shap_values(X_test_shap)

shap_importance = np.abs(shap_values).mean(axis=0)

shap.summary_plot(shap_values, X_test_shap,
                  feature_names=X_test_shap.columns,
                  plot_type="violin", max_display=25)

shap_imp_series = pd.Series(shap_importance, index=X_train_all.columns).sort_values(ascending=False)

# ══════════════════════════════════════════════════════════════════════════════
# ── 6. Sweep number of features ────────────────────────────────────────────────
# ══════════════════════════════════════════════════════════════════════════════
results = {}
n_feat_options = [10, 20, 30, 50, 70, 100, len(feature_cols)]

for n_feats in n_feat_options:
    top_feats = shap_imp_series.nlargest(n_feats).index.tolist()

    model = lgb.LGBMClassifier(**init_params, n_estimators=1000)
    model.fit(
        X_train_all[top_feats], Y_train_all,
        eval_set=[(X_test[top_feats], y_test)],
        eval_metric='binary_logloss',
        callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
    )

    prob  = model.predict_proba(X_test[top_feats])[:, 1]
    results[n_feats] = {
        'auc':      roc_auc_score(y_test, prob),
        'logloss':  log_loss(y_test, prob),
        'accuracy': accuracy_score(y_test, (prob > 0.5).astype(int)),
    }

print("\nFeature sweep results:")
print(f"  {'N':>5}  {'ROC-AUC':>8}  {'LogLoss':>8}  {'Accuracy':>9}")
for n, m in results.items():
    print(f"  {n:>5}  {m['auc']:>8.4f}  {m['logloss']:>8.4f}  {m['accuracy']*100:>8.2f}%")

# Plot the sweep
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
ns = list(results.keys())
axes[0].plot(ns, [results[n]['auc']      for n in ns], marker='o')
axes[0].set_xlabel('Number of features'); axes[0].set_ylabel('ROC-AUC'); axes[0].set_title('AUC vs Features')
axes[1].plot(ns, [results[n]['logloss']  for n in ns], marker='o', color='orange')
axes[1].set_xlabel('Number of features'); axes[1].set_ylabel('LogLoss'); axes[1].set_title('LogLoss vs Features')
plt.tight_layout(); plt.show()

# ── Pick best n_features from sweep ───────────────────────────────────────────
best_n    = max(results, key=lambda n: results[n]['auc'])
feat_shap = shap_imp_series.nlargest(best_n).index.tolist()
print(f"\nSelected {best_n} features (best AUC in sweep)")

# ══════════════════════════════════════════════════════════════════════════════
# ── 7. Optuna HP optimisation ──────────────────────────────────────────────────
# ══════════════════════════════════════════════════════════════════════════════
def objective(trial):
    lr      = trial.suggest_float("learning_rate", 1e-4, 0.05, log=True)
    patience = int(min(max(50, 0.1 / lr), 500))

    params = {
        "objective":         "binary",
        "boosting_type":     "gbdt",
        "verbosity":         -1,
        "random_state":      42,
        "n_jobs":            -1,
        "scale_pos_weight":  imbalance_ratio,
        "learning_rate":     lr,
        "num_leaves":        trial.suggest_int  ("num_leaves",        16,   256),
        "max_depth":         trial.suggest_int  ("max_depth",         3,    12),
        "min_child_samples": trial.suggest_int  ("min_child_samples", 5,    200),
        "subsample":         trial.suggest_float("subsample",         0.5,  1.0),
        "colsample_bytree":  trial.suggest_float("colsample_bytree",  0.5,  1.0),
        "reg_alpha":         trial.suggest_float("reg_alpha",         1e-8, 10.0, log=True),
        "reg_lambda":        trial.suggest_float("reg_lambda",        1e-8, 10.0, log=True),
    }

    model = lgb.LGBMClassifier(**params, n_estimators=10000)
    model.fit(
        X_train_all[feat_shap], Y_train_all,
        eval_set=[(X_test[feat_shap], y_test)],
        eval_metric='binary_logloss',
        callbacks=[lgb.early_stopping(patience), lgb.log_evaluation(0)],
    )

    prob = model.predict_proba(X_test[feat_shap])[:, 1]
    return roc_auc_score(y_test, prob)   # maximise AUC

study = optuna.create_study(
    direction="maximize",
    sampler=optuna.samplers.TPESampler(seed=42),
    pruner=optuna.pruners.MedianPruner(n_startup_trials=10, n_warmup_steps=2),
)
study.optimize(objective, n_trials=100, show_progress_bar=True)

print(f"\nBest ROC-AUC: {study.best_value:.4f}")
print("Best parameters:")
for k, v in study.best_params.items():
    print(f"  {k}: {v}")

# ══════════════════════════════════════════════════════════════════════════════
# ── 8. Final model with best params ───────────────────────────────────────────
# ══════════════════════════════════════════════════════════════════════════════
best_params = study.best_params
patience    = int(min(max(50, 0.1 / best_params['learning_rate']), 500))

lgb_train_final = lgb.Dataset(X_train_all[feat_shap], Y_train_all)
lgb_eval_final  = lgb.Dataset(X_test[feat_shap],      y_test, reference=lgb_train_final)

best_model = lgb.train(
    {**best_params, 'objective': 'binary', 'verbosity': -1,
     'scale_pos_weight': imbalance_ratio},
    lgb_train_final,
    num_boost_round=5000,
    valid_sets=lgb_eval_final,
    callbacks=[lgb.early_stopping(patience), lgb.log_evaluation(100)],
)

# ── Evaluate on held-out validation set ───────────────────────────────────────
val_prob  = best_model.predict(X_val[feat_shap], num_iteration=best_model.best_iteration)
val_pred  = (val_prob > 0.5).astype(int)

print("\nFinal model — validation set")
print(f"  ROC-AUC:  {roc_auc_score(y_val, val_prob):.4f}")
print(f"  Accuracy: {accuracy_score(y_val, val_pred)*100:.2f}%")
print(f"  LogLoss:  {log_loss(y_val, val_prob):.4f}")
print()
print(classification_report(y_val, val_pred,
      target_names=['Non-SC (tc=0)', 'Superconductor (tc>0)']))

# Confusion matrix
fig, ax = plt.subplots(figsize=(7, 5))
sns.heatmap(confusion_matrix(y_val, val_pred), annot=True, fmt='d',
            cmap='Blues', ax=ax,
            xticklabels=['Predicted Non-SC', 'Predicted SC'],
            yticklabels=['True Non-SC', 'True SC'])
ax.set_title(f'Confusion Matrix — Validation Set\nROC-AUC: {roc_auc_score(y_val, val_prob):.4f}')
plt.tight_layout(); plt.show()

# ── Probability histogram ──────────────────────────────────────────────────────
all_prob   = best_model.predict(df_wchem[feat_shap])
sc_prob    = all_prob[label.values == 1]
nonsc_prob = all_prob[label.values == 0]

plt.figure(figsize=(10, 7))
plt.hist(nonsc_prob, alpha=0.5, bins=50, label='Non-superconductor (tc=0)')
plt.hist(sc_prob,    alpha=0.3, bins=50, label='Superconductor (tc>0)')
plt.yscale('log')
plt.xlabel('Probability of superconductor')
plt.ylabel('Counts')
plt.title('All data on classification model')
plt.legend()
plt.tight_layout()
plt.savefig('classification_hist.png')
plt.show()

np.savetxt('SC_prob_all.txt', all_prob)
df_wchem.to_csv('df_wchem_classified.csv')
```
