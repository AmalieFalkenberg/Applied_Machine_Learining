```python
from __future__ import print_function, division
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import lightgbm as lgb
import shap
import optuna

from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, root_mean_squared_error
from sklearn.inspection import permutation_importance
from lightgbm import early_stopping

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "font.size": 16,
    "axes.grid": True,
    "grid.alpha": 0.2
})

# ── 1. Load data ───────────────────────────────────────────────────────────────
print("=== Loading Data ===\n")

df_wchem = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/df_wchem_icds.csv').drop('Unnamed: 0', axis=1)
tc       = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/tc_icds.csv').drop('Unnamed: 0', axis=1)
weights  = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/weights_icds.csv').drop('Unnamed: 0', axis=1)


mask = tc > 0
df_wchem = df_wchem[mask.values]
tc = tc[mask.values]

df_wchem = df_wchem.reset_index(drop=True)
tc       = tc.reset_index(drop=True)

print(f"Samples:  {len(tc)}")
print(f"Features: {df_wchem.shape[1]}")

# ── 2. Data quality ────────────────────────────────────────────────────────────
print("\n=== Data Quality Checks ===\n")

bool_cols = df_wchem.select_dtypes(include='bool').columns
df_wchem[bool_cols] = df_wchem[bool_cols].astype(int)

df_wchem = df_wchem.select_dtypes(include='number')

nan_total = df_wchem.isna().sum().sum()
if nan_total:
    print(f"Filling {nan_total} NaN values with column median")
    df_wchem = df_wchem.fillna(df_wchem.median())

zero_var = [c for c in df_wchem.columns if df_wchem[c].nunique() <= 1]
if zero_var:
    print(f"Dropping {len(zero_var)} zero-variance columns")
    df_wchem = df_wchem.drop(columns=zero_var)

print(f"✅ Clean feature matrix: {df_wchem.shape}")

# ── 3. Remove highly correlated features (>0.90) ──────────────────────────────
print("\n=== Removing Highly Correlated Features (threshold=0.90) ===\n")

corr_matrix  = df_wchem.corr(method='pearson').abs()
upper        = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop_corr = [c for c in upper.columns if any(upper[c] > 0.90)]
df_wchem     = df_wchem.drop(columns=to_drop_corr)

print(f"Dropped {len(to_drop_corr)} correlated features. Remaining: {df_wchem.shape[1]}")

# ── 4. Train / Test split (80 / 20) ───────────────────────────────────────────
# Val set is carved out of train only where needed (early stopping).
# The held-out test set is NEVER used for early stopping or feature selection.
print("\n=== Train / Test Split (80 / 20) ===\n")

X_train, Xp_test, y_train, yp_test = train_test_split(
    df_wchem, tc, test_size=0.40, random_state=42
)

X_test, X_val, y_test, y_val = train_test_split(
    Xp_test, yp_test, test_size=0.50, random_state=42
)

y_train = y_train.squeeze()
y_test  = y_test.squeeze()

y_train_log = np.log1p(y_train)
y_test_log  = np.log1p(y_test)

print(f"Train: {len(X_train)}  |  Test: {len(X_test)}")

# ── 5. Baseline model ──────────────────────────────────────────────────────────
# A small internal val split (from train only) is used purely for early stopping.
print("\n=== Baseline Model ===\n")

X_tr, X_es, y_tr, y_es = train_test_split(
    X_train, y_train_log, test_size=0.15, random_state=0
)

baseline_params = {
    'boosting_type': 'gbdt',
    'objective':     'regression',
    'num_leaves':    31,
    'learning_rate': 0.05,
    'n_estimators':  1000,
    'verbosity':     -1,
    'random_state':  42,
    'n_jobs':        -1,
}

gbm_base = lgb.LGBMRegressor(**baseline_params)
gbm_base.fit(
    X_tr, y_tr,
    eval_set=[(X_es, y_es)],       # early stopping uses only train-internal split
    eval_metric='rmse',
    callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
)

y_pred_base = np.expm1(gbm_base.predict(X_test))
rmse_base   = root_mean_squared_error(y_test, y_pred_base)
r2_base     = r2_score(y_test, y_pred_base)
print(f"Baseline — RMSE: {rmse_base:.3f} K   R²: {r2_base:.3f}  ({X_train.shape[1]} features)")

# ─────────────────────────────────────────────────────────────────────────────
# ── 6. FEATURE SELECTION — THREE METHODS (averaged over CV folds) ────────────
# ─────────────────────────────────────────────────────────────────────────────

print("\n" + "="*60)
print("  FEATURE SELECTION  (5-fold CV importance averaging)")
print("="*60)

N_FOLDS = 5
kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=42)

shap_accum = np.zeros(X_train.shape[1])
perm_accum = np.zeros(X_train.shape[1])
lgb_accum  = np.zeros(X_train.shape[1])

for fold, (tr_idx, va_idx) in enumerate(kf.split(X_train), 1):
    print(f"  Fold {fold}/{N_FOLDS} ...", end=' ', flush=True)

    Xf_tr, Xf_va = X_train.iloc[tr_idx], X_train.iloc[va_idx]
    yf_tr, yf_va = y_train_log.iloc[tr_idx], y_train_log.iloc[va_idx]

    # Internal early-stopping split (within this fold's train)
    Xf_tr2, Xf_es2, yf_tr2, yf_es2 = train_test_split(
        Xf_tr, yf_tr, test_size=0.15, random_state=fold
    )

    m = lgb.LGBMRegressor(**baseline_params)
    m.fit(
        Xf_tr2, yf_tr2,
        eval_set=[(Xf_es2, yf_es2)],
        eval_metric='rmse',
        callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
    )

    # SHAP
    exp   = shap.TreeExplainer(m)
    sv    = exp.shap_values(Xf_va.astype(float))
    shap_accum += np.abs(sv).mean(axis=0)

    # Permutation
    pr = permutation_importance(
        m, Xf_va, yf_va, n_repeats=5, random_state=fold, n_jobs=-1, scoring='r2'
    )
    perm_accum += pr.importances_mean

    # LGB gain
    lgb_accum += m.booster_.feature_importance(importance_type='gain')

    print("done")

# Average over folds
shap_importance = pd.Series(shap_accum / N_FOLDS, index=X_train.columns).sort_values(ascending=False)
perm_importance = pd.Series(perm_accum / N_FOLDS, index=X_train.columns).sort_values(ascending=False)
lgb_importance  = pd.Series(lgb_accum  / N_FOLDS, index=X_train.columns).sort_values(ascending=False)

print("\nTop 15 SHAP features (CV-averaged):")
for feat, val in shap_importance.head(15).items():
    print(f"  {feat:<50} {val:.5f}")

print("\nTop 15 Permutation importance features (CV-averaged):")
for feat, val in perm_importance.head(15).items():
    print(f"  {feat:<50} {val:.5f}")

print("\nTop 15 LGB gain importance features (CV-averaged):")
for feat, val in lgb_importance.head(15).items():
    print(f"  {feat:<50} {val:.1f}")

# ── 6a. Consensus feature set ─────────────────────────────────────────────────
N = 20
top_shap = set(shap_importance.head(N).index)
top_perm = set(perm_importance.head(N).index)
top_lgb  = set(lgb_importance.head(N).index)

all_candidates  = top_shap | top_perm | top_lgb
consensus_feats = sorted([
    f for f in all_candidates
    if sum([f in top_shap, f in top_perm, f in top_lgb]) >= 2
])

print(f"\nTop-{N} SHAP:        {len(top_shap)} features")
print(f"Top-{N} Permutation: {len(top_perm)} features")
print(f"Top-{N} LGB gain:    {len(top_lgb)} features")
print(f"Consensus (≥2/3):    {len(consensus_feats)} features")
print("\nConsensus feature list:")
for f in consensus_feats:
    in_methods = []
    if f in top_shap: in_methods.append("SHAP")
    if f in top_perm: in_methods.append("Perm")
    if f in top_lgb:  in_methods.append("LGB")
    print(f"  {f:<50} [{', '.join(in_methods)}]")

# ── 6b. Feature count sweep — CV-stabilised ───────────────────────────────────
print("\n--- Feature Count Sweep (5-fold CV, SHAP ranking) ---\n")

n_sweep = sorted(set(
    n for n in [5, 10, 15, 20, 25, 30, 40, 50, len(consensus_feats), X_train.shape[1]]
    if n <= X_train.shape[1]
))

sweep_results = {}

for n_feats in n_sweep:
    feats  = shap_importance.head(n_feats).index.tolist()
    m_cv   = lgb.LGBMRegressor(**baseline_params)
    scores = cross_val_score(
        m_cv,
        X_train[feats], y_train_log,
        cv=KFold(n_splits=5, shuffle=True, random_state=42),
        scoring='r2',
        n_jobs=-1,
    )
    # Also compute RMSE via cross_val_score with neg_root_mean_squared_error
    rmse_scores = cross_val_score(
        lgb.LGBMRegressor(**baseline_params),
        X_train[feats], y_train_log,
        cv=KFold(n_splits=5, shuffle=True, random_state=42),
        scoring='neg_root_mean_squared_error',
        n_jobs=-1,
    )
    sweep_results[n_feats] = {
        'r2_mean':   scores.mean(),
        'r2_std':    scores.std(),
        'rmse_mean': -rmse_scores.mean(),
        'rmse_std':  rmse_scores.std(),
    }

print(f"  {'N':>5}  RMSE (log)   ±std      R²      ±std")
print(f"  {'─'*55}")
for n, m in sweep_results.items():
    marker = ' ◄ baseline' if n == X_train.shape[1] else ''
    print(f"  {n:>5}  {m['rmse_mean']:>8.4f}   {m['rmse_std']:.4f}   {m['r2_mean']:.4f}   {m['r2_std']:.4f}{marker}")

# Plot sweep with error bars
fig, ax1 = plt.subplots(figsize=(10, 5))
ax2 = ax1.twinx()
ns        = list(sweep_results.keys())
rmse_m    = [sweep_results[n]['rmse_mean'] for n in ns]
rmse_s    = [sweep_results[n]['rmse_std']  for n in ns]
r2_m      = [sweep_results[n]['r2_mean']   for n in ns]
r2_s      = [sweep_results[n]['r2_std']    for n in ns]

ax1.errorbar(ns, rmse_m, yerr=rmse_s, fmt='b-o', capsize=4, label='RMSE (log-space)')
ax2.errorbar(ns, r2_m,   yerr=r2_s,   fmt='r-s', capsize=4, label='R²')
ax1.set_xlabel('Number of features')
ax1.set_ylabel('RMSE (log-space)', color='b')
ax2.set_ylabel('R²', color='r')
ax1.tick_params(axis='y', colors='b')
ax2.tick_params(axis='y', colors='r')
ax1.set_title('Feature count vs performance (5-fold CV, SHAP ranking)')
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='center right')
plt.tight_layout()
plt.savefig('feature_sweep.png', dpi=150, bbox_inches='tight')
plt.show()

# Pick best feature set
best_n    = 40 #min(sweep_results, key=lambda n: sweep_results[n]['rmse_mean'])
feat_shap = shap_importance.head(best_n).index.tolist()

# Prefer consensus if within 5% RMSE of best
if sweep_results.get(len(consensus_feats), {}).get('rmse_mean', 9e9) \
        <= sweep_results[best_n]['rmse_mean'] * 1.05:
    feat_shap = consensus_feats
    print(f"\n✅ Using consensus feature set ({len(feat_shap)} features)")
else:
    print(f"\n✅ Using top-{best_n} SHAP features (best CV RMSE)")

# SHAP summary plot on baseline model (trained on full train set)
print("\n=== SHAP Summary Plot (Baseline Model) ===\n")

gbm_base_full = lgb.LGBMRegressor(**baseline_params)
X_tr_f, X_es_f, y_tr_f, y_es_f = train_test_split(
    X_train, y_train_log, test_size=0.15, random_state=0
)
gbm_base_full.fit(
    X_tr_f, y_tr_f,
    eval_set=[(X_es_f, y_es_f)],
    eval_metric='rmse',
    callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
)

shap_test    = X_test.astype(float)
explainer_b  = shap.TreeExplainer(gbm_base_full)
shap_vals_b  = explainer_b.shap_values(shap_test)

shap.summary_plot(shap_vals_b, shap_test,
                  feature_names=shap_test.columns,
                  plot_type="violin", max_display=20, show=False)
plt.title("SHAP Feature Importance (Baseline Model)", fontsize=13)
plt.tight_layout()
plt.savefig('shap_summary_baseline.png', dpi=150, bbox_inches='tight')
plt.show()

# ─────────────────────────────────────────────────────────────────────────────
# ── 7. OPTUNA HYPERPARAMETER OPTIMISATION ────────────────────────────────────
# The objective uses a fresh internal val split (from train only).
# The test set is never touched during optimisation.
# ─────────────────────────────────────────────────────────────────────────────
print("\n=== Optuna Hyperparameter Optimisation ===\n")

# Fixed early-stopping split carved from train
X_opt_tr, X_opt_es, y_opt_tr, y_opt_es = train_test_split(
    X_train[feat_shap], y_train_log, test_size=0.15, random_state=7
)

def objective(trial):
    lr      = trial.suggest_float("learning_rate", 1e-4, 0.05, log=True)
    patience = int(min(max(50, 0.1 / lr), 500))

    params = {
        "objective":         "regression",
        "boosting_type":     "gbdt",
        "verbosity":         -1,
        "random_state":      42,
        "n_jobs":            -1,
        "learning_rate":     lr,
        "num_leaves":        trial.suggest_int  ("num_leaves",        16,  256),
        "max_depth":         trial.suggest_int  ("max_depth",          3,   12),
        "min_child_samples": trial.suggest_int  ("min_child_samples",  5,  200),
        "subsample":         trial.suggest_float("subsample",         0.5,  1.0),
        "colsample_bytree":  trial.suggest_float("colsample_bytree",  0.5,  1.0),
        "reg_alpha":         trial.suggest_float("reg_alpha",        1e-8, 10.0, log=True),
        "reg_lambda":        trial.suggest_float("reg_lambda",       1e-8, 10.0, log=True),
    }

    model = lgb.LGBMRegressor(**params, n_estimators=10000)
    model.fit(
        X_opt_tr, y_opt_tr,
        eval_set=[(X_opt_es, y_opt_es)],   # no test-set leakage
        eval_metric="rmse",
        callbacks=[lgb.early_stopping(patience), lgb.log_evaluation(0)],
    )
    # Evaluate on the internal val split (not the held-out test set)
    y_pred_es = np.expm1(model.predict(X_opt_es))
    y_true_es = np.expm1(y_opt_es)
    return root_mean_squared_error(y_true_es, y_pred_es)

optuna.logging.set_verbosity(optuna.logging.WARNING)
study = optuna.create_study(
    direction="minimize",
    sampler=optuna.samplers.TPESampler(seed=42),
    pruner=optuna.pruners.MedianPruner(n_startup_trials=10, n_warmup_steps=2),
)
study.optimize(objective, n_trials=100, show_progress_bar=True)

print(f"\nBest RMSE (internal val): {study.best_value:.3f} K")
print("Best parameters:")
for k, v in study.best_params.items():
    print(f"  {k}: {v}")

# ── 8. Retrain final model on full train set, evaluate on held-out test ────────
print("\n=== Final Model Evaluation ===\n")

best_p   = study.best_params
patience = int(min(max(50, 0.1 / best_p['learning_rate']), 500))

# For final training use an internal early-stopping split (not the test set)
X_fin_tr, X_fin_es, y_fin_tr, y_fin_es = train_test_split(
    X_train[feat_shap], y_train_log, test_size=0.15, random_state=99
)

lgb_train_ds = lgb.Dataset(X_fin_tr, y_fin_tr)
lgb_es_ds    = lgb.Dataset(X_fin_es, y_fin_es, reference=lgb_train_ds)

full_params = {
    **best_p,
    'objective':     'regression',
    'boosting_type': 'gbdt',
    'verbosity':     -1,
    'random_state':  42,
}

gbm_final = lgb.train(
    full_params,
    lgb_train_ds,
    num_boost_round=5000,
    valid_sets=lgb_es_ds,
    callbacks=[early_stopping(patience), lgb.log_evaluation(0)],
)

# Final evaluation on the held-out test set (first time it's used since baseline)
y_pred_final = np.expm1(gbm_final.predict(X_test[feat_shap]))
rmse_final   = root_mean_squared_error(y_test, y_pred_final)
mae_final    = mean_absolute_error(y_test, y_pred_final)
r2_final     = r2_score(y_test, y_pred_final)

print(f"Test performance ({len(feat_shap)} features):")
print(f"  RMSE : {rmse_final:.4f} K")
print(f"  MAE  : {mae_final:.4f} K")
print(f"  R²   : {r2_final:.4f}")
print(f"\nImprovement over baseline:")
print(f"  RMSE : {rmse_base:.3f} → {rmse_final:.3f} K  ({(rmse_base-rmse_final)/rmse_base*100:+.1f}%)")
print(f"  R²   : {r2_base:.3f} → {r2_final:.3f}")

# ── 9. Plots ───────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(13, 5), dpi=100)

for ax, y_pred, title in zip(
    axes,
    [y_pred_base, y_pred_final],
    [f'Baseline ({X_train.shape[1]} feats)  R²={r2_base:.3f}',
     f'Optimised ({len(feat_shap)} feats)  R²={r2_final:.3f}']
):
    ax.scatter(y_test, y_pred, s=2, alpha=0.5)
    lims = [min(y_test.min(), np.array(y_pred).min()),
            max(y_test.max(), np.array(y_pred).max())]
    ax.plot(lims, lims, 'r--', lw=1)
    ax.set_xlabel('True Tc (K)')
    ax.set_ylabel('Predicted Tc (K)')
    ax.set_title(title, fontsize=13)
    ax.set_xscale('log')
    ax.set_yscale('log')

plt.suptitle('Tc Regression — Baseline vs Optimised', fontsize=14)
plt.tight_layout()
plt.savefig('regression_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

# ── 10. SHAP on final model ────────────────────────────────────────────────────
print("\n=== SHAP Summary (Final Model) ===\n")

explainer_final = shap.TreeExplainer(gbm_final)
shap_final      = explainer_final.shap_values(X_test[feat_shap].astype(float))

shap.summary_plot(shap_final, X_test[feat_shap].astype(float),
                  feature_names=feat_shap,
                  plot_type="violin", max_display=20, show=False)
plt.title("SHAP Feature Importance (Optimised Model)", fontsize=13)
plt.tight_layout()
plt.savefig('shap_summary_final.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nSaved: shap_summary_baseline.png, feature_sweep.png,")
print("       regression_comparison.png, shap_summary_final.png")
```

```python
# ══════════════════════════════════════════════════════════════════════════════
# 0. Imports
# ══════════════════════════════════════════════════════════════════════════════
from __future__ import print_function, division
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import lightgbm as lgb
import shap
import optuna

from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.metrics import (
    mean_squared_error, mean_absolute_error, r2_score,
    root_mean_squared_error
)
from sklearn.inspection import permutation_importance

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "font.size": 16,
    "axes.grid": True,
    "grid.alpha": 0.2
})

# ══════════════════════════════════════════════════════════════════════════════
# 1. Load data — superconductors only
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Loading Data ===\n")

df_wchem = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/df_wchem_icds.csv').drop('Unnamed: 0', axis=1)
tc       = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/tc_icds.csv').drop('Unnamed: 0', axis=1)
weights  = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/weights_icds.csv').drop('Unnamed: 0', axis=1)

tc_series = tc.squeeze().reset_index(drop=True)
df_wchem  = df_wchem.reset_index(drop=True)
weights   = weights.squeeze().reset_index(drop=True)


# Keep only superconductors (Tc > 0)
mask = tc_series > 0
df_wchem = df_wchem[mask.values].reset_index(drop=True)
tc       = tc_series[mask.values].reset_index(drop=True)
weights  = weights[mask.values].reset_index(drop=True)


print(f"Superconducting samples: {len(tc)}")
print(f"Features: {df_wchem.shape[1]}")

# ══════════════════════════════════════════════════════════════════════════════
# 2. Data cleaning
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Data Cleaning ===\n")

# Convert bool → int
bool_cols = df_wchem.select_dtypes(include="bool").columns
df_wchem[bool_cols] = df_wchem[bool_cols].astype(int)

# Keep numeric only
df_wchem = df_wchem.select_dtypes(include="number")

# Fill NaN
if df_wchem.isna().sum().sum() > 0:
    df_wchem = df_wchem.fillna(df_wchem.median())

# Remove zero‑variance
zero_var = [c for c in df_wchem.columns if df_wchem[c].nunique() <= 1]
if zero_var:
    df_wchem = df_wchem.drop(columns=zero_var)

print(f"After cleaning: {df_wchem.shape}")

# ══════════════════════════════════════════════════════════════════════════════
# 3. Remove highly correlated features (>0.90)
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Removing Highly Correlated Features ===\n")

corr = df_wchem.corr().abs()
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
to_drop = [c for c in upper.columns if any(upper[c] > 0.90)]

df_wchem = df_wchem.drop(columns=to_drop)
print(f"Dropped {len(to_drop)} correlated features. Remaining: {df_wchem.shape[1]}")

# ══════════════════════════════════════════════════════════════════════════════
# 4. Train / Test split (80 / 20) — NO STRATIFY
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Train / Test Split (80/20, no stratify) ===\n")

X_train, X_test, y_train, y_test, w_train, w_test = train_test_split(
    df_wchem, tc, weights, test_size=0.20, random_state=42
)

y_train_log = np.log1p(y_train)
y_test_log  = np.log1p(y_test)

print(f"Train: {len(X_train)}  |  Test: {len(X_test)}")

# ══════════════════════════════════════════════════════════════════════════════
# 5. Baseline model
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Baseline Model ===\n")

X_tr, X_es, y_tr, y_es, w_tr, w_es = train_test_split(
    X_train, y_train_log, w_train, test_size=0.15, random_state=0
)

baseline_params = {
    "boosting_type": "gbdt",
    "objective": "regression",
    "num_leaves": 31,
    "learning_rate": 0.05,
    "n_estimators": 1000,
    "verbosity": -1,
    "random_state": 42,
    "n_jobs": -1,
}

gbm_base = lgb.LGBMRegressor(**baseline_params)
gbm_base.fit(
    X_tr, y_tr,
    sample_weight=w_tr,
    eval_set=[(X_es, y_es)],
    eval_sample_weight=[w_es],
    eval_metric="rmse",
    callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
)

y_pred_base = np.expm1(gbm_base.predict(X_test))
rmse_base   = root_mean_squared_error(y_test, y_pred_base)
r2_base     = r2_score(y_test, y_pred_base)

print(f"Baseline RMSE: {rmse_base:.3f} K")
print(f"Baseline R²:   {r2_base:.3f}")

# ══════════════════════════════════════════════════════════════════════════════
# 6. Feature importance (SHAP + Permutation + LGB gain)
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Feature Importance (5-fold CV) ===\n")

N_FOLDS = 5
kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=42)

shap_accum = np.zeros(X_train.shape[1])
perm_accum = np.zeros(X_train.shape[1])
lgb_accum  = np.zeros(X_train.shape[1])

for fold, (tr_idx, va_idx) in enumerate(kf.split(X_train), 1):
    print(f"Fold {fold}/{N_FOLDS}")

    Xf_tr, Xf_va = X_train.iloc[tr_idx], X_train.iloc[va_idx]
    yf_tr, yf_va = y_train_log.iloc[tr_idx], y_train_log.iloc[va_idx]
    wf_tr, wf_va = w_train.iloc[tr_idx], w_train.iloc[va_idx]

    # Early stopping split inside fold
    Xf_tr2, Xf_es2, yf_tr2, yf_es2, wf_tr2, wf_es2 = train_test_split(
        Xf_tr, yf_tr, wf_tr, test_size=0.15, random_state=fold
    )

    m = lgb.LGBMRegressor(**baseline_params)
    m.fit(
        Xf_tr2, yf_tr2,
        sample_weight=wf_tr2,
        eval_set=[(Xf_es2, yf_es2)],
        eval_sample_weight=[wf_es2],
        eval_metric="rmse",
        callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
    )

    # SHAP
    exp = shap.TreeExplainer(m)
    sv  = exp.shap_values(Xf_va.astype(float))
    shap_accum += np.abs(sv).mean(axis=0)

    # Permutation
    pr = permutation_importance(
        m, Xf_va, yf_va,
        n_repeats=5, random_state=fold,
        sample_weight=wf_va, scoring="r2", n_jobs=-1
    )
    perm_accum += pr.importances_mean

    # LGB gain
    lgb_accum += m.booster_.feature_importance(importance_type="gain")

# Average
shap_importance = pd.Series(shap_accum / N_FOLDS, index=X_train.columns).sort_values(ascending=False)
perm_importance = pd.Series(perm_accum / N_FOLDS, index=X_train.columns).sort_values(ascending=False)
lgb_importance  = pd.Series(lgb_accum  / N_FOLDS, index=X_train.columns).sort_values(ascending=False)

# Consensus
N = 20
top_shap = set(shap_importance.head(N).index)
top_perm = set(perm_importance.head(N).index)
top_lgb  = set(lgb_importance.head(N).index)

consensus_feats = sorted([
    f for f in (top_shap | top_perm | top_lgb)
    if sum([f in top_shap, f in top_perm, f in top_lgb]) >= 2
])

print(f"Consensus features: {len(consensus_feats)}")

# ══════════════════════════════════════════════════════════════════════════════
# 7. Feature count sweep (CV)
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Feature Count Sweep ===\n")

n_sweep = [5, 10, 15, 20, 30, 40, 50, len(consensus_feats), X_train.shape[1]]
n_sweep = sorted(set([n for n in n_sweep if n <= X_train.shape[1]]))

sweep_results = {}

for n_feats in n_sweep:
    feats = shap_importance.head(n_feats).index.tolist()
    model = lgb.LGBMRegressor(**baseline_params)

    scores = cross_val_score(
        model,
        X_train[feats], y_train_log,
        cv=KFold(n_splits=5, shuffle=True, random_state=42),
        scoring="neg_root_mean_squared_error",
        n_jobs=-1,
    )

    sweep_results[n_feats] = -scores.mean()

best_n = min(sweep_results, key=sweep_results.get)
feat_shap = shap_importance.head(best_n).index.tolist()

print(f"Selected {best_n} SHAP features")

# ══════════════════════════════════════════════════════════════════════════════
# 8. Optuna hyperparameter tuning
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Optuna Hyperparameter Optimisation ===\n")

X_opt_tr, X_opt_es, y_opt_tr, y_opt_es, w_opt_tr, w_opt_es = train_test_split(
    X_train[feat_shap], y_train_log, w_train, test_size=0.15, random_state=7
)

def objective(trial):
    lr = trial.suggest_float("learning_rate", 1e-4, 0.05, log=True)
    patience = int(min(max(50, 0.1 / lr), 500))

    params = {
        "objective": "regression",
        "boosting_type": "gbdt",
        "verbosity": -1,
        "random_state": 42,
        "n_jobs": -1,
        "learning_rate": lr,
        "num_leaves": trial.suggest_int("num_leaves", 16, 256),
        "max_depth": trial.suggest_int("max_depth", 3, 12),
        "min_child_samples": trial.suggest_int("min_child_samples", 5, 200),
        "subsample": trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.5, 1.0),
        "reg_alpha": trial.suggest_float("reg_alpha", 1e-8, 10.0, log=True),
        "reg_lambda": trial.suggest_float("reg_lambda", 1e-8, 10.0, log=True),
    }

    model = lgb.LGBMRegressor(**params, n_estimators=10000)
    model.fit(
        X_opt_tr, y_opt_tr,
        sample_weight=w_opt_tr,
        eval_set=[(X_opt_es, y_opt_es)],
        eval_sample_weight=[w_opt_es],
        eval_metric="rmse",
        callbacks=[lgb.early_stopping(patience), lgb.log_evaluation(0)],
    )

    pred = np.expm1(model.predict(X_opt_es))
    true = np.expm1(y_opt_es)
    return root_mean_squared_error(true, pred)

optuna.logging.set_verbosity(optuna.logging.WARNING)
study = optuna.create_study(direction="minimize")
study.optimize(objective, n_trials=100, show_progress_bar=True)

best_params = study.best_params
print("\nBest parameters:")
for k, v in best_params.items():
    print(f"  {k}: {v}")

# ══════════════════════════════════════════════════════════════════════════════
# 9. Final model training + evaluation
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Final Model Evaluation ===\n")

X_fin_tr, X_fin_es, y_fin_tr, y_fin_es, w_fin_tr, w_fin_es = train_test_split(
    X_train[feat_shap], y_train_log, w_train, test_size=0.15, random_state=99
)

patience = int(min(max(50, 0.1 / best_params["learning_rate"]), 500))

train_ds = lgb.Dataset(X_fin_tr, y_fin_tr, weight=w_fin_tr)
es_ds    = lgb.Dataset(X_fin_es, y_fin_es, weight=w_fin_es, reference=train_ds)

final_params = {
    **best_params,
    "objective": "regression",
    "boosting_type": "gbdt",
    "verbosity": -1,
    "random_state": 42,
}

gbm_final = lgb.train(
    final_params,
    train_ds,
    num_boost_round=5000,
    valid_sets=es_ds,
    callbacks=[lgb.early_stopping(patience), lgb.log_evaluation(0)],
)

# Evaluate on untouched test set
y_pred_final = np.expm1(gbm_final.predict(X_test[feat_shap]))
rmse_final   = root_mean_squared_error(y_test, y_pred_final)
mae_final    = mean_absolute_error(y_test, y_pred_final)
r2_final     = r2_score(y_test, y_pred_final)

print(f"Final RMSE: {rmse_final:.3f} K")
print(f"Final MAE:  {mae_final:.3f} K")
print(f"Final R²:   {r2_final:.3f}")

# ══════════════════════════════════════════════════════════════════════════════
# 10. SHAP summary (final model)
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== SHAP Summary (Final Model) ===\n")

explainer_final = shap.TreeExplainer(gbm_final)
shap_vals_final = explainer_final.shap_values(X_test[feat_shap].astype(float))

shap.summary_plot(
    shap_vals_final,
    X_test[feat_shap].astype(float),
    feature_names=feat_shap,
    plot_type="violin",
    max_display=20
)
```

```python
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "font.size": 16,
    "axes.grid": True,
    "grid.alpha": 0.2
})

# ══════════════════════════════════════════════════════════════════════════════
# 1. Load data — superconductors only
# ══════════════════════════════════════════════════════════════════════════════
ref      = pd.read_csv('/Users/fbnielsen/Desktop/3DSC/superconductors_3D/data/final/ICSD/3DSC_ICSD.csv',
                       skiprows=1)['database_code_icsd_2']
df_wchem = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/df_wchem_icds.csv').drop('Unnamed: 0', axis=1)
tc       = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/tc_icds.csv').drop('Unnamed: 0', axis=1)
weights  = pd.read_csv('/Users/fbnielsen/Desktop/Initial_class_reg_dimred/3DSC/Dimensionality_reduction/weights_icds.csv').drop('Unnamed: 0', axis=1)  # FIX: axiz → axis

tc_series = tc.squeeze().reset_index(drop=True)
df_wchem  = df_wchem.reset_index(drop=True)
weights   = weights.squeeze().reset_index(drop=True)
ref       = ref.squeeze().reset_index(drop=True)

# Keep only superconductors (Tc > 0)
mask     = tc_series > 0
df_wchem = df_wchem[mask.values].reset_index(drop=True)
tc       = tc_series[mask.values].reset_index(drop=True)
weights  = weights[mask.values].reset_index(drop=True)
ref      = ref[mask.values].reset_index(drop=True)

print(f"Superconducting samples: {len(tc)}")
print(f"Features: {df_wchem.shape[1]}")

# ══════════════════════════════════════════════════════════════════════════════
# 2. Data cleaning
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Data Cleaning ===\n")

bool_cols = df_wchem.select_dtypes(include="bool").columns
df_wchem[bool_cols] = df_wchem[bool_cols].astype(int)
df_wchem  = df_wchem.select_dtypes(include="number")

if df_wchem.isna().sum().sum() > 0:
    df_wchem = df_wchem.fillna(df_wchem.median())

zero_var = [c for c in df_wchem.columns if df_wchem[c].nunique() <= 1]
if zero_var:
    df_wchem = df_wchem.drop(columns=zero_var)

print(f"After cleaning: {df_wchem.shape}")

# ══════════════════════════════════════════════════════════════════════════════
# 3. Remove highly correlated features (> 0.90)
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Removing Highly Correlated Features ===\n")

corr  = df_wchem.corr().abs()
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
to_drop = [c for c in upper.columns if any(upper[c] > 0.90)]
df_wchem = df_wchem.drop(columns=to_drop)
print(f"Dropped {len(to_drop)} correlated features. Remaining: {df_wchem.shape[1]}")

# ══════════════════════════════════════════════════════════════════════════════
# 4. Train / Val / Test split (60 / 20 / 20) grouped by ICSD
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Train / Val / Test Split (60/20/20) ===\n")

from sklearn.model_selection import GroupShuffleSplit, KFold, train_test_split

groups = ref.values

gss1 = GroupShuffleSplit(n_splits=1, test_size=0.3, random_state=42)
train_idx, temp_idx = next(gss1.split(df_wchem, groups=groups))

gss2 = GroupShuffleSplit(n_splits=1, test_size=0.5, random_state=42)
temp_groups  = groups[temp_idx]
test_rel_idx, val_rel_idx = next(gss2.split(df_wchem.iloc[temp_idx], groups=temp_groups))

test_idx = temp_idx[test_rel_idx]
val_idx  = temp_idx[val_rel_idx]

X_train      = df_wchem.iloc[train_idx]
Y_train      = tc.iloc[train_idx]
weight_train = weights.iloc[train_idx].squeeze()

X_val        = df_wchem.iloc[val_idx]
Y_val        = tc.iloc[val_idx]
weight_val   = weights.iloc[val_idx].squeeze()

X_test       = df_wchem.iloc[test_idx]
Y_test       = tc.iloc[test_idx]              # FIX: consistent capitalisation
weight_test  = weights.iloc[test_idx].squeeze()

y_train_log = np.log1p(Y_train)
y_val_log   = np.log1p(Y_val)
y_test_log  = np.log1p(Y_test)

print(f"Train: {len(X_train)}  |  Test: {len(X_test)}  |  Val: {len(X_val)}")  # FIX: X_train_all → X_train

# ══════════════════════════════════════════════════════════════════════════════
# 5. Baseline model
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Baseline Model ===\n")

baseline_params = {
    "boosting_type":  "gbdt",
    "objective":      "regression",
    "num_leaves":     31,
    "learning_rate":  0.05,
    "n_estimators":   1000,
    "verbosity":      -1,
    "random_state":   42,
    "n_jobs":         -1,
}

gbm_base = lgb.LGBMRegressor(**baseline_params)
gbm_base.fit(
    X_train, y_train_log,               # train on log scale
    sample_weight=weight_train,
    eval_set=[(X_val, y_val_log)],
    eval_sample_weight=[weight_val],
    eval_metric="rmse",
    callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
)

y_pred_base = np.expm1(gbm_base.predict(X_test))
rmse_base   = root_mean_squared_error(Y_test, y_pred_base)  # FIX: y_test → Y_test
r2_base     = r2_score(Y_test, y_pred_base)

print(f"Baseline RMSE: {rmse_base:.3f} K")
print(f"Baseline R²:   {r2_base:.3f}")

# ══════════════════════════════════════════════════════════════════════════════
# 6. Feature importance (SHAP + Permutation + LGB gain) — 5-fold CV
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Feature Importance (5-fold CV) ===\n")

N_FOLDS   = 5
kf        = KFold(n_splits=N_FOLDS, shuffle=True, random_state=42)

shap_accum = np.zeros(X_train.shape[1])
perm_accum = np.zeros(X_train.shape[1])
lgb_accum  = np.zeros(X_train.shape[1])

for fold, (t_idx, v_idx) in enumerate(kf.split(X_train), 1):
    print(f"Fold {fold}/{N_FOLDS}")

    Xf_tr, Xf_va = X_train.iloc[t_idx],      X_train.iloc[v_idx]
    yf_tr, yf_va = y_train_log.iloc[t_idx],   y_train_log.iloc[v_idx]   # FIX: Y_train_log → y_train_log
    wf_tr, wf_va = weight_train.iloc[t_idx],  weight_train.iloc[v_idx]

    Xf_tr2, Xf_es2, yf_tr2, yf_es2, wf_tr2, wf_es2 = train_test_split(
        Xf_tr, yf_tr, wf_tr, test_size=0.15, random_state=fold
    )

    m = lgb.LGBMRegressor(**baseline_params)
    m.fit(
        Xf_tr2, yf_tr2,
        sample_weight=wf_tr2,
        eval_set=[(Xf_es2, yf_es2)],
        eval_sample_weight=[wf_es2],
        eval_metric="rmse",
        callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)],
    )

    # SHAP
    sv = shap.TreeExplainer(m).shap_values(Xf_va.astype(float))
    shap_accum += np.abs(sv).mean(axis=0)

    # Permutation
    pr = permutation_importance(
        m, Xf_va, yf_va,
        n_repeats=5, random_state=fold,
        sample_weight=wf_va, scoring="r2", n_jobs=-1
    )
    perm_accum += pr.importances_mean

    # LGB gain
    lgb_accum += m.booster_.feature_importance(importance_type="gain")

shap_importance = pd.Series(shap_accum / N_FOLDS, index=X_train.columns).sort_values(ascending=False)
perm_importance = pd.Series(perm_accum / N_FOLDS, index=X_train.columns).sort_values(ascending=False)
lgb_importance  = pd.Series(lgb_accum  / N_FOLDS, index=X_train.columns).sort_values(ascending=False)

# Consensus: features in at least 2 of 3 top-20 lists
N        = 20
top_shap = set(shap_importance.head(N).index)
top_perm = set(perm_importance.head(N).index)
top_lgb  = set(lgb_importance.head(N).index)

consensus_feats = sorted([
    f for f in (top_shap | top_perm | top_lgb)
    if sum([f in top_shap, f in top_perm, f in top_lgb]) >= 2
])
print(f"Consensus features (≥2 methods): {len(consensus_feats)}")

# ══════════════════════════════════════════════════════════════════════════════
# 7. Feature count sweep — manual CV to support sample_weight
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Feature Count Sweep ===\n")

n_sweep = sorted(set([5, 10, 15, 20, 30, 40, 50,
                       len(consensus_feats), X_train.shape[1]]))
n_sweep = [n for n in n_sweep if n <= X_train.shape[1]]

sweep_results = {}

for n_feats in n_sweep:
    feats  = shap_importance.head(n_feats).index.tolist()
    fold_rmses = []

    for t_idx, v_idx in KFold(n_splits=5, shuffle=True, random_state=42).split(X_train):
        Xs_tr, Xs_va = X_train[feats].iloc[t_idx], X_train[feats].iloc[v_idx]
        ys_tr, ys_va = y_train_log.iloc[t_idx],     y_train_log.iloc[v_idx]
        ws_tr        = weight_train.iloc[t_idx]

        m = lgb.LGBMRegressor(**baseline_params)
        m.fit(Xs_tr, ys_tr, sample_weight=ws_tr,
              callbacks=[lgb.log_evaluation(0)])

        pred = np.expm1(m.predict(Xs_va))
        true = np.expm1(ys_va)
        fold_rmses.append(root_mean_squared_error(true, pred))

    sweep_results[n_feats] = np.mean(fold_rmses)
    print(f"  n={n_feats:4d}  CV-RMSE={sweep_results[n_feats]:.3f} K")

best_n    = min(sweep_results, key=sweep_results.get)
feat_shap = shap_importance.head(best_n).index.tolist()
print(f"\nSelected {best_n} features (lowest CV-RMSE)")

# ══════════════════════════════════════════════════════════════════════════════
# 8. Optuna — uses X_train for fitting, X_val for early stopping + scoring
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Optuna Hyperparameter Optimisation ===\n")

def objective(trial):
    lr       = trial.suggest_float("learning_rate", 1e-4, 0.05, log=True)
    patience = int(min(max(50, 0.1 / lr), 500))

    params = {
        "objective":         "regression",
        "boosting_type":     "gbdt",
        "verbosity":         -1,
        "random_state":      42,
        "n_jobs":            -1,
        "learning_rate":     lr,
        "num_leaves":        trial.suggest_int  ("num_leaves",        16,  256),
        "max_depth":         trial.suggest_int  ("max_depth",         3,   12),
        "min_child_samples": trial.suggest_int  ("min_child_samples", 5,   200),
        "subsample":         trial.suggest_float("subsample",         0.5, 1.0),
        "colsample_bytree":  trial.suggest_float("colsample_bytree",  0.5, 1.0),
        "reg_alpha":         trial.suggest_float("reg_alpha",         1e-8, 10.0, log=True),
        "reg_lambda":        trial.suggest_float("reg_lambda",        1e-8, 10.0, log=True),
    }

    model = lgb.LGBMRegressor(**params, n_estimators=10000)
    model.fit(
        X_train[feat_shap], y_train_log,       # train on selected features
        sample_weight=weight_train,
        eval_set=[(X_val[feat_shap], y_val_log)],  # early stop on val
        eval_sample_weight=[weight_val],
        eval_metric="rmse",
        callbacks=[lgb.early_stopping(patience), lgb.log_evaluation(0)],
    )

    # Score on val — same set used for early stopping is fine here
    # since we're not selecting hyperparams based on a separate holdout;
    # Optuna's TPE sampler handles the exploration/exploitation tradeoff
    pred = np.expm1(model.predict(X_val[feat_shap]))
    true = np.expm1(y_val_log)
    return root_mean_squared_error(true, pred)

optuna.logging.set_verbosity(optuna.logging.WARNING)
study = optuna.create_study(direction="minimize")
study.optimize(objective, n_trials=100, show_progress_bar=True)

best_params = study.best_params
print("\nBest parameters:")
for k, v in best_params.items():
    print(f"  {k}: {v}")

# ══════════════════════════════════════════════════════════════════════════════
# 9. Final model — train on train, early-stop on val, evaluate on test
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== Final Model Evaluation ===\n")

patience = int(min(max(50, 0.1 / best_params["learning_rate"]), 500))

# FIX: all datasets use feat_shap consistently
train_ds = lgb.Dataset(X_train[feat_shap], y_train_log,  weight=weight_train)
es_ds    = lgb.Dataset(X_val[feat_shap],   y_val_log,    weight=weight_val,
                        reference=train_ds)

gbm_final = lgb.train(
    {**best_params, "objective": "regression",
     "boosting_type": "gbdt", "verbosity": -1, "random_state": 42},
    train_ds,
    num_boost_round=5000,
    valid_sets=es_ds,
    callbacks=[lgb.early_stopping(patience), lgb.log_evaluation(100)],
)

# Test set — never seen during training or tuning
y_pred_final = np.expm1(gbm_final.predict(X_test[feat_shap]))
print(f"Test  RMSE: {root_mean_squared_error(Y_test, y_pred_final):.3f} K")
print(f"Test  MAE:  {mean_absolute_error(Y_test, y_pred_final):.3f} K")
print(f"Test  R²:   {r2_score(Y_test, y_pred_final):.3f}")

# Val set — used for early stopping, so slightly optimistic but standard practice
y_pred_val = np.expm1(gbm_final.predict(X_val[feat_shap]))
print(f"\nVal   RMSE: {root_mean_squared_error(Y_val, y_pred_val):.3f} K")
print(f"Val   MAE:  {mean_absolute_error(Y_val, y_pred_val):.3f} K")
print(f"Val   R²:   {r2_score(Y_val, y_pred_val):.3f}")

# ══════════════════════════════════════════════════════════════════════════════
# 10. SHAP summary (final model)
# ══════════════════════════════════════════════════════════════════════════════
print("\n=== SHAP Summary (Final Model) ===\n")

explainer_final = shap.TreeExplainer(gbm_final)
shap_vals_final = explainer_final.shap_values(X_test[feat_shap].astype(float))

shap.summary_plot(
    shap_vals_final,
    X_test[feat_shap].astype(float),
    feature_names=feat_shap,
    plot_type="violin",
    max_display=20
)

### RUN ME!
```
