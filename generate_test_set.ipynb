```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pymatgen.core import Structure
```

```python
data_mp = pd.read_csv("MP/3DSC_MP.csv", skiprows=1)
data_ICSD = pd.read_csv("3DSC_ICSD.csv", skiprows=1)
```

```python
data_mp.loc[0,"cif"].replace("data/final/","")
```

```python
icsd_formulas = set(data_ICSD["formula_sc"])

tc = []
cif_loc = []

for i, row in data_mp.iterrows():
    if row["formula_sc"] not in icsd_formulas:
        tc.append(row["tc"])
        cif_loc.append(row["cif"].replace("data/final/",""))
        print(f"{len(tc)}", end="\r")

print(f"\nFound {len(tc)} MP-only structures")

data_reduced = pd.DataFrame({"tc": tc, "cif": cif_loc})
data_reduced.to_csv("MP/MP_only_structures.csv", index=False)
```
